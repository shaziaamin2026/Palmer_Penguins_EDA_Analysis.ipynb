Correlation Analysis

Correlation analysis measures the relationship
between numerical variables.

It helps us identify whether two variables
have a positive, negative, or weak relationship.

I use the corr() function to calculate
correlation coefficients.

A heatmap is used to visualize these relationships.

```python
# Select numerical columns
numeric_df = df_clean.select_dtypes(include=np.number)

# Calculate correlation
correlation = numeric_df.corr()

# Visualize correlation
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f")

plt.title("Penguin Correlation Heatmap")

plt.show()
```

Outlier Detection Using IQR

Outliers are values that are unusually
different from other observations.

The Interquartile Range (IQR) method
helps us identify potential outliers.

IQR = Q3 - Q1

Where:
- Q1 = First Quartile.
- Q3 = Third Quartile.

I use a boxplot to visualize
potential outliers in penguin body mass.

```python
Q1 = df_clean["body_mass_g"].quantile(0.25)
Q3 = df_clean["body_mass_g"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df_clean[(df_clean["body_mass_g"] < lower_limit)  | (df_clean["body_mass_g"] > upper_limit)]

print("Total Outliers:", len(outliers))

display(outliers)
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df_clean,x="species",y="body_mass_g")

plt.title("Penguin Body Mass Outlier Analysis")

plt.show()
```
